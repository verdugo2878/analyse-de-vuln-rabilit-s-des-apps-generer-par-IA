## Étape 3 : Les quatre familles de textes
Rappel de la séance :

| Famille | Exemple | Obligatoire ? | Qui vérifie ? |
|---|---|---|---|
| **Loi** | Loi 25, RGPD | oui, si elle s'applique à nous | une autorité (CAI, CNIL…), avec des amendes |
| **Cadre** | NIST CSF, CIS Controls | non : on le choisit pour s'organiser | personne, sauf nous |
| **Norme** | ISO 27001 | non par la loi, mais souvent **exigée par contrat** | un auditeur tiers, qui certifie |
| **Standard** | MITRE ATT&CK | non : un accord informel de l'industrie | personne |

Attention : « obligatoire » dépend **des faits**. Le RGPD est une loi européenne, et il peut pourtant s'appliquer à une PME de Saguenay.

## Étape 4 : Texte par texte
**Ce qu'on fait :** on écrit les faits de l'entreprise, puis une **règle** par texte. Chaque règle renvoie l'applicabilité (`oui`, `non` ou `volontaire`) et une justification d'une phrase.
Écrire les règles dans du code oblige à être précis, et permet de tout recalculer si un fait change.

````python
# <-- MODIFIABLE : les faits de l'entreprise (tirés de la fiche)
faits = {
    'siege_au_quebec': True,
    'renseignements_personnels': True,       # clients, employés, candidats
    'offre_services_residents_ue': True,     # site en français, prix en euros, 431 clients en France
    'etablissement_dans_ue': False,
    'employes': 35,
    'chiffre_affaires_millions_euros': 4.2,
    'donnees_sante_americaines': False,
    'clients_californie': 0,
    'contrat_exige_iso27001': True,
}

def loi25(f):
    if f['siege_au_quebec'] and f['renseignements_personnels']:
        return 'oui', "Entreprise du Québec qui traite des renseignements personnels (clients, employés, candidats)."
    return 'non', "Pas d'activité au Québec impliquant des renseignements personnels."

def rgpd(f):
    if f['offre_services_residents_ue'] or f['etablissement_dans_ue']:
        return 'oui', "Offre ses services à des personnes en France (art. 3(2) du RGPD), même sans établissement dans l'UE."
    return 'non', "Aucune activité visant des personnes dans l'UE."

def hipaa(f):
    return ('oui', "Traite des données de santé américaines.") if f['donnees_sante_americaines'] \
        else ('non', "Aucune donnée de santé ni entité de santé américaine.")

def ccpa(f):
    return ('oui', "A des clients en Californie.") if f['clients_californie'] > 0 \
        else ('non', "Aucun client en Californie.")

def iso27001(f):
    return ('oui', "Pas exigée par la loi, mais exigée par contrat par la Banque Saguenéenne.") if f['contrat_exige_iso27001'] \
        else ('volontaire', "Aucun client ne l'exige ; certification possible par choix.")

textes = [   # texte, famille, règle
    ('Loi 25', 'loi', loi25),
    ('RGPD', 'loi', rgpd),
    ('HIPAA', 'loi', hipaa),
    ('CCPA', 'loi', ccpa),
    ('ISO 27001', 'norme', iso27001),
    ('NIST CSF 2.0', 'cadre', lambda f: ('volontaire', "Utilisé par choix dans le rapport annuel à la direction.")),
    ('CIS Controls v8', 'cadre', lambda f: ('volontaire', "L'équipe TI suit les CIS Controls v8 (IG1) par choix.")),
    ('MITRE ATT&CK', 'standard', lambda f: ('volontaire', "Référence de l'industrie pour décrire les attaques ; personne ne l'impose.")),
]
````

## Ce qu'il faut retenir
- Une **loi** s'impose si ses conditions sont remplies (lieu, données, clients), même si elle vient d'un autre pays : le RGPD s'applique à une PME de Saguenay qui sert des clients en France.
- Un **cadre** (NIST, CIS) et un **standard** (MITRE ATT&CK) sont volontaires : on les choisit pour s'organiser.
- Une **norme** (ISO 27001) n'est pas une loi, mais un contrat peut la rendre obligatoire. C'est un tiers qui vérifie et certifie.
- L'applicabilité se **justifie par des faits** : si un fait change (taille, clients), la liste change.
